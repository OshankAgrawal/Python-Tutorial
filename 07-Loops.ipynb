## `range()` function

It is an iterable.

range() functio has 3 argument.

1. start
2. stop
3. step

`range(1, 6, 2)` -> here start = 1 , stop = 6 , step = 2

`range(2, 6)` ->  here start = 2, stop = 6 , step = 1 (by default)

`range(6)` -> here start = 0 (by default), stop = 6, step = 1 (by default)

**Note**: In the range function we have to pass atleast one argument.

```python
for i in range(1, 6):
    print(i)
```

```python
for i in range(10):
    print(i)
```

```python
for i in range(1, 10, 2):
    print(i)
```

```python
for i in range(-1, -10, -2):
    print(i)
```

```python
for i in range():
    print(i)
```

```python
for i in range('a', 'z'):
    print(i)
```

```python
for i in range(1.5, 10):
    print(i)
```

```python
for i in range(1, 10, 1.5):
    print(i)
```

## 1. for loop

```python
for i in [1, 2, 3, 4, 5, 6]:
    print(i)
```

```python
for i in (1, 2, 3, 4, 5, 6):
    print(i)
```

```python
for i in 'Hello World':
    print(i)
```

```python
dict = {
    'name': 'Oshank',
    'age': 25
}

for x in dict:  # It return only keys
    print(x)

for a, b in dict.items(): # It return both keys and values
    print(a," ",b)
```

```python
for i in [1, 2, 3, 4, 5, 6]:
    i *= 2
    print(i)
```

```python
for i in {1, 2, 3, 4, 5, 6}:
    i *= 2
    print(i)
```

```python
for i in range(1, 10):
    print(i)
```

```python
for i in range(10):
    print('Oshank')
```

## 2. While loop

```python
# Code complexity increase

print('Oshank')
print('Oshank')
print('Oshank')
print('Oshank')
print('Oshank')
print('Oshank')
print('Oshank')
print('Oshank')
print('Oshank')
print('Oshank')
```

```python
i = 0
while i < 10:
    print('Oshank')
    i += 2
```

```python
i = 0
while i < 100:
    print(i, end=" ")
    i += 2
```

```python
i = 0
while i < 100:
    print(i)
    
```

## Nested Loop

```python
## Nested loopss
## a loop inside a loop

for i in range(3): # 0, 1, 2
    for j in range(4): # 0, 1, 2, 3
        print("i =", i, " and j =", j)
```

```python
num = int(input("Input a number"))

for i in range(1, 11, 1):
    print(num," * ",i,"= ", num*i)
```
