# Exercise 4 (Loops)

Question 01: Program to print table of a number, say 5.

Question 02: Program to print sum of natural numbers between 1 to 7. Print the sum progressively, i.e., after adding each natural number, print sum so far.

Question 03: Program to calculate and print the sums of even and odd integers of the first n natural numbers.

Question 04: Program to input some numbers repeatedly and print their sum. The program ends when the users say no more to enter (normal termination) or program aborts when the number entered is less than zero.

Question 5: Program to input a number and test if it is a prime number.
