## 2. Defining Function in Python

You can create functions in a program, that :
- can have arguments (values given to it), if needed
- can perform certain functionality (some set of statements)
- can return a result

```python
## syntax
def function_name(parameters):
    """Docstring"""
    # Function body
    return expression
```

where:
- `def` means a **function definition** is starting
- `identifier following 'def'` is the **name of the function**
- the `variables/identifiers inside the parentheses` are the **arguments** or **parameters** (values given to function)
- there is a `colon` at the end of def line, meaning **it requires a block**
- the statements indented below the function, (i.e., block below def line) define the functionality (working) of the function. This block is also called body-of-the-function.
- The return statement returns the computed result.

```python
def calculate_cube(x):
    return x**3
```

## 5. Returning values from function

Functions in python may or may not return a value. 

Based on it function in python classify in two types:
1. Non-void functions (Functions returning some value)
2. Void functions (Functions not returning any value)

### 5.1 Non-void functions (Functions returning some value)

The functions that return some computed result in terms of a value, fall in this category.\
The computed value is returned using `return` keyword.\
Syntax: `return <value>`

Note: The value being returned can be one of the following:
- a literal
- a variable
- an expression

**The return statements ends a function execution even if it is in middle of the function.**\
A function ends the moment when it reaches a return statement or all statements in function-body have been executed, whichever occurs earlier.

**Example: Non-void Function – Calculate Student's Percentage**

A non-void function is a function that returns a value to the calling statement.

Let's take a real-life example where we calculate a student's percentage based on marks obtained in five subjects.

```python
def calculate_percentage(marks):
    total = sum(marks)
    percentage = (total / 500) * 100
    return percentage


student_marks = [85, 90, 78, 88, 95]

result = calculate_percentage(student_marks)

print("Student Percentage:", result)
```

### 5.2 Void functions (Functions not returning any value)

The functions that perform some action or do some work but do not return any computed value or final value to the caller are called void function.\
A void function may or may not have a return statement, if a void function has a return statement, then it takes the following form:\
    `return`

that is, keyword **return** without any value or expression.

```python
def greet():
    print("Hello")

a = greet()
print(f"returned value: {a}")
```

So, now you know that in Python you can have following four possible combinations of functions :
1. non-void functions without any arguments
2. non-void functions with some arguments
3. void functions without any arguments
4. void functions with some arguments

Please note that a function in a program can call any other function in the same program.

## 6. Function Composition

Composition in general refers to using an expression as part of a larger expression; or a statement as a part of larger statement.\
In function's context, we can understand composition as follows:\
The arguments of a function call can be any kind of expression :
- an arithmelic expression e.g ..\
`greater((4+5), (3+4))`
- a logical expression e.g ..\
`test( a or b)`
- a function call (function composition) eg.\
`int(str(52))`\
`int(float("52.5") *2)`\
`int(str(52) +str(10))`

The above examples show you composition of function calls
- a function call as part of larger function call.

Note: Composition In general refers to using an expression as part of a larger expression; or a statement as a part of larger statement.

#### 1. Arithmetic Expression as an Argument

```python
def greater(a, b):
    if a > b:
        return a
    else:
        return b

result = greater((4 + 5), (3 + 4))
print(result)
```

#### 2. Logical Expression as an Argument

```python
def test(value):
    if value:
        print("Condition is True")
    else:
        print("Condition is False")

a = 10
b = 20

test(a > b or a < b)
```

#### 3. Function Call as Part of Another Function Call (Function Composition)

Example 1: `int(str(52))`

```python
result = int(str(52))
print(result)
print(type(result))
```

Example 2: `int(float("52.5") * 2)`

```python
result = int(float("52.5") * 2)
print(result)
```

Example 3: `int(str(52) + str(10))`

```python
result = int(str(52) + str(10))
print(result)
```

### The fundamental rule:
Python always evaluates the expressions inside a function call before passing their resulting values to the function.\
In function composition, the innermost function call is evaluated first, and its result becomes the input to the outer function.

## 7. Arguments and Parameters

### 7.1 Arguments

Arguments are the actual values that we pass to a function when calling it.

### 7.2 Parameters

Parameters are the **variables defined inside the parentheses of a function** definition. They act as placeholders for the values that the function will receive when it is called.

From a function's perspective:
- A **parameter** is the **variable listed inside the parentheses** in the function definition.
- An *argument* is the **actual value that is sent to the function** when it is called.

```python
def greet(name):  # name is a parameter
    print("Hello,", name)
```

```python
greet("Oshank") # "Oshank" is an argument
```

Note: Parameters are the variables that receive values, whereas arguments are the values being passed.

**Default Parameter Values**

You can assign default values to parameters. If the function is called without an argument, it uses the default value:

```python
def my_function(country = "India"): # country is a parameter with default value
  print("I am from", country)

my_function("Sweden")
my_function("USA")
my_function()
my_function("Brazil")
```

**Keyword Arguments**

You can send arguments with the key = value syntax.\
The phrase Keyword Arguments is often shortened to `kwargs` in Python documentation.

```python
def my_function(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

my_function(animal = "dog", name = "Buddy")
```

This way, with keyword arguments, the order of the arguments does not matter.

```python
def my_function(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

my_function(name = "Buddy", animal = "dog")
```

**Positional Arguments**

When you call a function with arguments without using keywords, they are called positional arguments.

Positional arguments must be in the correct order:

```python
def my_function(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

my_function("dog", "Buddy")
```

The order matters with positional arguments:

```python
# Switching the order changes the result:

def my_function(animal, name):
  print("I have a", animal)
  print("My", animal + "'s name is", name)

my_function("Buddy", "dog")
```

Mixing Positional and Keyword Arguments
You can mix positional and keyword arguments in a function call.

However, positional arguments must come before keyword arguments:

```python
def my_function(animal, name, age):
  print("I have a", age, "year old", animal, "named", name)

my_function("dog", name = "Buddy", age = 5)
```

Passing Different Data Types
You can send any data type as an argument to a function (string, number, list, dictionary, etc.).

The data type will be preserved inside the function:

```python
# Sending a list as an argument:

def my_function(fruits):
  for fruit in fruits:
    print(fruit)

my_fruits = ["apple", "banana", "cherry"]
my_function(my_fruits)
```

```python
# Sending a dictionary as an argument:

def my_function(person):
  print("Name:", person["name"])
  print("Age:", person["age"])

my_person = {"name": "Emil", "age": 25}
my_function(my_person)
```

Positional-Only Arguments
You can specify that a function can have ONLY positional arguments.

To specify positional-only arguments, add , / after the arguments:

```python
def my_function(name, /):
  print("Hello", name)

my_function("Emil")
```

Without the , / you are actually allowed to use keyword arguments even if the function expects positional arguments:

```python
def my_function(name):
  print("Hello", name)

my_function(name = "Emil")
```

With , /, you will get an error if you try to use keyword arguments:

```python
def my_function(name, /):
  print("Hello", name)

my_function(name = "Emil")
```

Keyword-Only Arguments
To specify that a function can have only keyword arguments, add *, before the arguments:

```python
def my_function(*, name):
  print("Hello", name)

my_function(name = "Emil")
```

Without *,, you are allowed to use positional arguments even if the function expects keyword arguments:

```python
def my_function(name):
  print("Hello", name)

my_function("Emil")
```

With *,, you will get an error if you try to use positional arguments:

```python
def my_function(*, name):
  print("Hello", name)

my_function("Emil")
```

Combining Positional-Only and Keyword-Only
You can combine both argument types in the same function.

Arguments before / are positional-only, and arguments after * are keyword-only:

```python
def my_function(a, b, /, *, c, d):
  return a + b + c + d

result = my_function(5, 10, c = 15, d = 20)
print(result)
```

## 8. Python *args and **kwargs

By default, a function must be called with the correct number of arguments.

However, sometimes you may not know how many arguments that will be passed into your function.

*args and **kwargs allow functions to accept a unknown number of arguments.

Arbitrary Arguments - *args
If you do not know how many arguments will be passed into your function, add a * before the parameter name.
Arbitrary Arguments are often shortened to *args in Python documentation.
This way, the function will receive a tuple of arguments and can access the items accordingly:

```python
# Using *args to accept any number of arguments:

def my_function(*kids):
  print("The youngest child is " + kids[2])

my_function("Emil", "Tobias", "Linus")
```

What is *args?
The *args parameter allows a function to accept any number of positional arguments.

Inside the function, args becomes a tuple containing all the passed arguments:

```python
# Accessing individual arguments from *args:

def my_function(*args):
  print("Type:", type(args))
  print("First argument:", args[0])
  print("Second argument:", args[1])
  print("All arguments:", args)

my_function("Emil", "Tobias", "Linus")
```

Using *args with Regular Arguments
You can combine regular parameters with *args.

Regular parameters must come before *args:

```python
def my_function(greeting, *names):
  for name in names:
    print(greeting, name)

my_function("Hello", "Emil", "Tobias", "Linus")
```

In this example, "Hello" is assigned to greeting, and the rest are collected in names.

Practical Example with *args
*args is useful when you want to create flexible functions:

```python
# A function that calculates the sum of any number of values:

def my_function(*numbers):
  total = 0
  for num in numbers:
    total += num
  return total

print(my_function(1, 2, 3))
print(my_function(10, 20, 30, 40))
print(my_function(5))
```

```python
# Finding the maximum value:

def my_function(*numbers):
  if len(numbers) == 0:
    return None
  max_num = numbers[0]
  for num in numbers:
    if num > max_num:
      max_num = num
  return max_num

print(my_function(3, 7, 2, 9, 1))
```

Arbitrary Keyword Arguments - **kwargs
If you do not know how many keyword arguments will be passed into your function, add two asterisks ** before the parameter name.

This way, the function will receive a dictionary of arguments and can access the items accordingly:

```python
# Using **kwargs to accept any number of keyword arguments:

def my_function(**kid):
  print("His last name is " + kid["lname"])

my_function(fname = "Tobias", lname = "Refsnes")
```

What is **kwargs?
The **kwargs parameter allows a function to accept any number of keyword arguments.

Inside the function, kwargs becomes a dictionary containing all the keyword arguments:
Arbitrary Keyword Arguments are often shortened to **kwargs in Python documentation.

```python
# Accessing values from **kwargs:

def my_function(**myvar):
  print("Type:", type(myvar))
  print("Name:", myvar["name"])
  print("Age:", myvar["age"])
  print("All data:", myvar)

my_function(name = "Tobias", age = 30, city = "Bergen")
```

Using **kwargs with Regular Arguments
You can combine regular parameters with **kwargs.

Regular parameters must come before **kwargs:

```python
def my_function(username, **details):
  print("Username:", username)
  print("Additional details:")
  for key, value in details.items():
    print(" ", key + ":", value)

my_function("emil123", age = 25, city = "Oslo", hobby = "coding")
```

Combining *args and **kwargs
You can use both *args and **kwargs in the same function.

The order must be:

regular parameters
*args
**kwargs

```python
def my_function(title, *args, **kwargs):
  print("Title:", title)
  print("Positional arguments:", args)
  print("Keyword arguments:", kwargs)

my_function("User Info", "Emil", "Tobias", age = 25, city = "Oslo")
```

Unpacking Arguments
The * and ** operators can also be used when calling functions to unpack (expand) a list or dictionary into separate arguments.

Unpacking Lists with *
If you have values stored in a list, you can use * to unpack them into individual arguments:

```python
# Using * to unpack a list into arguments:

def my_function(a, b, c):
  return a + b + c

numbers = [1, 2, 3]
result = my_function(*numbers) # Same as: my_function(1, 2, 3)
print(result)
```

Unpacking Dictionaries with **
If you have keyword arguments stored in a dictionary, you can use ** to unpack them:

```python
# Using ** to unpack a dictionary into keyword arguments:

def my_function(fname, lname):
  print("Hello", fname, lname)

person = {"fname": "Emil", "lname": "Refsnes"}
my_function(**person) # Same as: my_function(fname="Emil", lname="Refsnes")
```

**Remember: Use * and ** in function definitions to collect arguments, and use them in function calls to unpack arguments.**

## 9. Scope of a Variable

The scope of a variable determines where that variable can be accessed or used in a program.\
In Python, a variable is not necessarily accessible everywhere. Its accessibility depends on where it is defined.

Python mainly has four types of scopes:
1. Local Scope
2. Enclosing Scope
3. Global Scope
4. Built-in Scope

### 9.1 Local Scope

A variable defined inside a function is called a local variable.\
It can generally be accessed only inside that function.

```python
def display():
    name = "Rahul"
    print(name)

display()
```

Here, name is a local variable because it is defined inside the display() function.

**What happens if we access it outside the function?**

```python
def display():
    name = "Rahul"
    print(name)

display()

print(name)
```

Because name exists within the local scope of display(). It cannot be accessed directly from outside that function.

### 9.2 Global Scope

A variable defined outside all functions is called a global variable.\
It can be accessed from different parts of the program, including inside functions.

```python
name = "Rahul"

def display():
    print(name)

display()
print(name)
```

Here, name is a global variable because it is defined outside the function.

The function display() can access it.

#### Modifying a Global Variable Inside a Function

```python
balance = 1000

def deposit():
    balance = 500
    print("Inside function:", balance)

deposit()
print("Outside function:", balance)
```

Notice that the global balance remains 1000.

Why?

Because when we write:
balance = 500


inside the function, Python creates a new local variable named balance rather than modifying the global variable.
To modify a global variable, we use the global keyword.

```python
balance = 1000

def deposit():
    global balance
    balance = balance + 500

deposit()

print(balance)
```

The global keyword tells Python that we want to refer to the global variable balance, rather than creating a new local variable.

### 9.3 Enclosing Scope

Enclosing scope occurs when a function is defined inside another function.\
The inner function can access variables defined in its outer (enclosing) function.

```python
def outer():
    name = "Rahul"

    def inner():
        print(name)

    inner()

outer()
```

Here:
- outer() is the outer function.
- inner() is the inner function.
- name is defined inside outer().
- inner() can access name because it is in the enclosing scope.

**Enclosing vs. Local Scope**

```python
def outer():
    name = "Rahul"

    def inner():
        name = "Amit"
        print("Inner:", name)

    inner()
    print("Outer:", name)

outer()
```

Why?

The inner function creates its own local variable name.

Therefore, it doesn't modify the name variable of the outer function.

**Modifying an Enclosing Variable Using nonlocal**

If we want to modify the variable of an enclosing function, we can use the nonlocal keyword.

```python
def outer():
    balance = 1000

    def inner():
        nonlocal balance
        balance += 500

    inner()
    print(balance)

outer()
```

The nonlocal keyword tells Python to use the variable from the enclosing function's scope instead of creating a new local variable.

### 9.4 Built-in Scope

Python has some predefined names that are available without defining them ourselves.\
These names belong to the built-in scope.

Examples:
- `print()`
- `len()`
- `sum()`
- `max()`
- `min()`
- `range()`
- `int()`
- `str()`

We can use these functions directly because they are available in Python's built-in scope.

```python
numbers = [10, 20, 30]

print(len(numbers))
print(sum(numbers))
print(max(numbers))
```

Important points to remember
- A variable's scope determines where it can be accessed.
- Variables defined inside a function are generally local to that function.
- Variables defined outside functions are global.
- Inner functions can access variables from their enclosing functions.
- global allows us to modify a global variable inside a function.
- nonlocal allows us to modify a variable in an enclosing function.
- Python follows the LEGB rule to search for variables.

LEGB Rule
When Python tries to find a variable, it searches in the following order:
| Order | Scope | Meaning |
|---|---|---|
| L | Local | Current function |
| E | Enclosing | Outer function(s) |
| G | Global | Module/program level |
| B | Built-in | Python predefined names |


Python searches for a variable in this order:

Local → Enclosing → Global → Built-in

As soon as Python finds the variable, it stops searching.

Example Demonstrating LEGB

```python
name = "Global"

def outer():
    name = "Enclosing"

    def inner():
        name = "Local"
        print(name)

    inner()

outer()
```

Python finds name in the local scope of inner() first, so it doesn't search the enclosing, global, or built-in scopes.

**What if the local variable doesn't exist?**

```python
name = "Global"

def outer():
    name = "Enclosing"

    def inner():
        print(name)

    inner()

outer()
```

Python searches:
1. Local scope → Not found
2. Enclosing scope → Found "Enclosing"
Therefore, it prints "Enclosing".

**In Python, multiple functions can have variables with the same name, but those variables may refer to different memory locations and have different scopes.**

1. Same Local Variable Name in Multiple Functions

```python
def function1():
    name = "Rahul"
    print("Function 1:", name)


def function2():
    name = "Amit"
    print("Function 2:", name)


function1()
function2()
```

Explanation
Here, both functions have a variable named name.

However:
- name inside function1() is a local variable of function1().
- name inside function2() is a local variable of function2().
Although their names are identical, they are two independent local variables.

Changing one does not affect the other.

2. Same Global Variable Accessed by Multiple Functions

```python
balance = 10000

def function1():
    print("Function 1:", balance)


def function2():
    print("Function 2:", balance)


function1()
function2()
print("Outside:", balance)
```

Explanation
Here, balance is defined outside both functions.

Therefore:
- function1() accesses the global balance.
- function2() accesses the same global balance.
- The statement outside the functions also accesses the same global balance.
All three access the same variable.

3. Same Variable Name: One Global and One Local

```python
balance = 10000

def function1():
    balance = 5000
    print("Function 1:", balance)


def function2():
    print("Function 2:", balance)


function1()
function2()
print("Outside:", balance)
```

Explanation
Let's understand what happens.

Step 1: Python creates a global variable.
balance = 10000


Step 2: Python calls function1().
def function1():    balance = 5000    print("Function 1:", balance)


When Python encounters balance = 5000, it creates a new local variable because the assignment occurs inside the function.

Now we have:
Variable	Scope	Value
balance	Global	10000
balance	Local to function1	5000


Step 3: Python calls function2().
def function2():    print("Function 2:", balance)


There is no local variable named balance inside function2().

Therefore, Python searches the global scope and finds balance = 10000.

Step 4: Finally, Python executes:
print("Outside:", balance)


It accesses the global variable, which is still 10000.

Important: Creating a local variable with the same name as a global variable does not modify the global variable.

4. Modifying the Same Global Variable from Multiple Functions

What if we actually want to modify the global variable from different functions?\
We can use the global keyword.

```python
balance = 10000

def deposit():
    global balance
    balance = balance + 5000
    print("After Deposit:", balance)


def withdraw():
    global balance
    balance = balance - 2000
    print("After Withdrawal:", balance)


deposit()
withdraw()

print("Final Balance:", balance)
```

Explanation
Initially:
Global balance = 10000

When deposit() executes:
balance = balance + 5000


The global balance becomes:
Global balance = 15000

Next, withdraw() executes:
balance = balance - 2000


The same global variable is modified:
Global balance = 13000

Finally, the global balance is 13000.

5. Comparison of All Four Cases

| Situation | Function 1 | Function 2 | Do they share the same variable? |
|---|---|---|---|
| Both have local variables with the same name | Local `name` | Local `name` | No |
| Both access the same global variable | Global `balance` | Global `balance` | Yes |
| One has a local variable, another accesses global | Local `balance` | Global `balance` | No |
| Both modify the global variable using `global` | Global `balance` | Global `balance` | Yes |
