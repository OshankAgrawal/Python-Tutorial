## 1. Creating a Dictionary

In this method all the `key: value` pairs of a dictionary are written collectively, separated by commas and enclosed in **curly braces** as per following syntax :

`<dictionary-name> = {<key> :< value>, <key> :< value> ... }`

```python
teachers = {
    "Dimple": "Computer Science",
    "Karen": "Sociology",
    "Harpreet": "Mathematics",
    "Sabah" : "Legal Studies"
}

print(type(teachers))
print(teachers)
```

Note Points:

2. Each entry **(Key: Value)** consists of a pair separated by a colon - the key and corresponding value is given by writing colon (:) between them.
3. The key-value pairs are separated by commas (,).

## 1.4 Creating a Dictionary from `name and value Pairs`

Using the `dict()` constructor of dictionary, you can also create a new dictionary initialized from specified set of keys and values. There are multiple ways to provide keys and values to `dict()` constructor.

#### 1.4.1. Specify `Key: Value` pairs as keyword **arguments** to `dict()` function, Keys as arguments and Values as their values.\
Here the argument names are taken as keys of String type.

```python
Employee = dict(name ='John', salary = 10000, age = 24)

print(Employee)
```

#### 1.4.2. Specify comma-separated `key: value` pairs

To give `key: value` pairs in this format, you need to enclose them in **curly braces** as shown below :

```python
Employee = dict({'name' : 'John', 'salary' : 10000, 'age' : 24})
print(Employee)
```

#### 1.4.3 Specify keys separately and corresponding values separately

In this method, the `key` and `values` are enclosed separately in `parentheses` and are given as arguments to the `zip()`
function, which is then given as argument of `dict()`.

```python
keys = ('name', 'salary', 'age')
values = ('John', 10000, 24)

Employee = dict(zip(keys, values) )

print(Employee)
```

Note points:

1. All the keys are given separatety enclosed in () and all the values are given separately enclosed in () as arguments to zip function. The zip() function is given as argument of dict() constructor.

2. First group contain all the keys and second group contains all values in same order as that of corresponding keys.

3. The zip function clubs first value from first set with the first value of second set, second value from
first set with the second value of second set, and so on.

#### 1.4.4 Specify Key: Value pairs separately in form of sequences

In this method, one list or tuple argument is passed to `dict()`. This argument (passed list or tuple) contains lists/tuples of individual `key: value` pairs. That is, a `key: value` pair is specified in the form of a list and there are as many lists as items of the outer list as there are `key: value` pairs in the dictionary.

```python
Employee = dict([['name', 'John'], ['salary ', 10000], ['age', 24]])

print(Employee)
```

NOTE: Important. Using dict( ) method to create a dictionary takes longer time compared to traditional method of encosing values in {} Thus, it should be avoided.

## 5. Characteristics of a Dictionary

Dictionaries like lists are mutable and that is the only similarity they have with lists.\
Otherwise, dictionaries are different type of data structures with following characteristics :

1. Unordered Set
2. Not a Sequence
3. Indexed by Keys, Not Numbers
4. Keys must be Unique
5. Mutable
6. Internally Stored as Mappings

### 5.1. Unordered Set

A dictionary is a unordered set of key : value pairs. Its values can contain references to any type of object.

### 5.2. Not a Sequence

Unlike a string, list and tuple, a dictionary is not a sequence because it is unordered set of elements. The sequences are indexed by a range of ordinal numbers. Hence, they are ordered but a dictionary is an unordered collection.

### 5.3. Indexed by Keys, Not Numbers

Dictionaries are indexed by keys. According to Python, a key can be "any non-mutable type".\
Since strings and numbers are not mutable, you can use them as a key in a dictionary.\
And if a tuple contains immutable objects such as integers or strings etc., then only it can also be used as key.\
But the values in a dictionary can be of any type, and types can be mixed within one.

```python
dict1 ={
    0 : "Value for key @",
    1 : "Value for key 1",
    "3" : "Value for a string-as-a-key",
    (4, 5): "Value for a tuple-as-a-key",
    "and for fun" : 7
}
```

Note: The keys of a dictionary must be of immutable types such as numbers or strings. A tuple can also be a key provided it contains only immutable elements.

### 5.4. Keys must be Unique

Each of the keys within a dictionary must be unique. Since keys are used to identify values in a dictionary, there cannot be duplicate keys in a dictionary.\
However, two unique keys can have same values, e.g., consider the same BirdCount dictionary.

```python
dis = {
    1: "1",
    2: "2",
    3: "3",
    4: "4"
}

print(dis)

dis[5] = "5"

print(dis)

dis[4] = "6"

print(dis)
```

```python
BirdCount
```

Notice that two different keys "Hornbill" and "Peacock" have same value 15.

### 5.5. Mutable

Like lists, dictionaries are also mutable. We can change the value of a certain key in place using the assignment statement as per syntax : `<dictionary>[<key>] = <value>`

```python
dis = {
    1: "1",
    2: "2",
    3: "3",
    4: "4"
}

print(dis)

dis[5] = "5"

print(dis)

dis[4] = "6"

print(dis)
```

You can even add a new key:value pair to a dictionary using a simple assignment statement, But the key being added should be unique. If the key already exists, then value is simply changed.

### 5.6. Internally Stored as Mappings

Internally, the key: value pairs of a dictionary are associated with one another with some internal function (called hash-function). This way of linking is called mapping.

### 6.3 Deleting element from a dictionary

There are two methods for deleting elements from a dictionary.

#### 6.3.1  using del keyword

To delete a dictionary element or a dictionary entry, i.e., a key:value pair, you can use del
command.

Syntax: `del <dictionary>[<key>]`

Note: But with del statement, the key that you are giving to delete must exist in the dictionary, otherwise Python raises exception (KeyError).

```python
emp = {'salary' : 10000, 'age' : 24, 'name' : 'John'}

print(emp)

del emp['age']

print(emp)
```

```python
emp = {'salary' : 10000, 'age' : 24, 'name' : 'John'}

print(emp)

del emp['city']

print(emp)
```

#### 6.3.2 using pop() method

The `pop()` method will not only delete the `key:value` pair for mentioned key but also **return the corresponding value**.

Syntax : `<dictionary>.pop(<key>)`

Note: If you try to delete a key which does not exist, the Python returns error.

```python
employee={'salary' : 10000, 'age' : 24, 'name' : 'John'}
print(employee)

x = employee.pop('age')
print(x)

print(employee)
```

```python
employee={'salary' : 10000, 'age' : 24, 'name' : 'John'}
print(employee)

x = employee.pop('city')
print(x)

print(employee)
```

However, `pop()` method allows you to specify what to display when the given key does not exist, rather than the default error message. This can be done as per following syntax : `<dictionary>.pop(<key>, <in-case-of-error-show-me>)`

```python
employee={'salary' : 10000, 'age' : 24, 'name' : 'John'}
print(employee)

x = employee.pop('city', "Not Found")
print(x)

print(employee)
```

See, now Python returned the specified message in place of error-message.

### 6.4 Checking for Existence of a Key

Usual membership operators `in` and `not in` work with dictionaries as well. But they can **check for the existence of keys only**. To check for whether a value is present in a dictionary (called reverse lookup), you need to write proper code for that.

To use a membership operator for a key's presence in a dictionary, you may write statement as per syntax given below :

`<key> in <dictionary>`
`<key> not in <dictionary>`

- The `in` operator will **return True** if the **given key is present** in the dictionary, **otherwise False**.
- The `not in` operator will **return True** if the given **key is not present** in the dictionary, **otherwise False**.

```python
empl={'salary' : 10000, 'age' : 24, 'name' : 'John'}

'age' in empl
```

```python
'John' in empl
```

```python
'John' not in empl
```

```python
'age' not in empl
```

### 6.6 Counting Frequency of Elements in a list using Dictionary

Since a dictionary uses key:value pairs to store its elements, you can use this feature effectively if you trying to calculate frequency of elements in sequence such as lists.

You can do this as :
- Create an empty dictionary
- Take up an element from the list List (first element 1st time, second element 2nd time and so on)
- Check if this element exists as a key in the dictionary :
If not then add {key:value} to dictionary in the form { List-element : count of List-element in List}

```python
import json
sentence = "This is a super idea This idea will change the idea of learning"
words = sentence.split()

d = {}

for x in words:
    if x not in d:
        d[x] = words.count(x)

print("Count frequencies in list \n", words)
print(json.dumps(d, indent=4))
```

### 7.3 `get()` method

With this method, you can **get the item with the given key**, similar to `dictionary[<key>]` If the key is not present, Python by default gives error, but you can specify your own message through default argument as per following syntax :\
`<dictionary>.get( key, [ default ])`

- Takes key as essential argument and returns corresponding value if the key exists.
- `dis[<key>]` (using square brackets) gives a KeyError if the key is missing.
- `dis.get(<key>)` (using .get()) **returns None by default** and never raises an error. That is the main reason developers use `.get()` —to safely look up keys without crashing the program.

```python
dis = {'salary' : 10000, 'dept' : 'Sales', 'age': 24, 'name' : 'John'}

print(dis.get('salary'))
```

```python
print(dis.get('city', ['This is not a key in disctionary.']))
```

```python
print(dis.get('city'))
```

### 7.4 `items()` method

This method returns all of the items in the dictionary as a sequence of (key, value) tuples.\
Note that these are returned in no particular order.\
Syntax: `<dictionary>.items()`

- Takes no argument ; Returns a sequence of (key, value) pairs

```python
employee = {'name' : 'John', 'salary' : 10000, 'age' : 24}
myList = employee.items()
for x in myList :
    print(x)
```

As the items() function returns a sequence of (key value) pairs, you can write a loop having two variables to access key value pairs, e.g.,

```python
empl={'age': 25, 'name': 'Bhuvan', 'salary': 20000}
seq= empl.items()
for key, val in seq:
    print(val, key)
```

### 7.7 `update()` method

This method merges key: value pairs from the new dictionary into the original dictionary, adding or replacing as needed. The items in the new dictionary are added to the old one and override any items already there with the same keys.\
Syntax: `<dictionary>.update ( <other-dictionary>)`

- This dictionary's items will be taken for updating other dictionary.

```python
employee1 = {'name' : 'John', 'salary' : 10000, 'age' : 24}
employee2= {'name' : 'Diya','salary' : 54000, 'dept' : 'Sales'}

print(f"First Dictionary: {employee1}")
print(f"Second Dictionary: {employee2}")

employee1.update(employee2)

print(f"First Dictionary: {employee1}")
print(f"Second Dictionary: {employee2}")
```
