## DataTypes
#### 1. Definition:

- Data types are a classification of data which tell the compiler or interpreter how the programmer intends to use the data.
- They determine the type of operations that can be performed on the data, the values that the data can take, and the amount of memory needed to store the data.

#### 2. Importance of Data Types in Programming
Explanation:

- Data types ensure that data is stored in an efficient way.
- They help in performing correct operations on data.
- Proper use of data types can prevent errors and bugs in the program.

### 6. List

When we want to assing or store multiple values to a single variable.

List is defined with the square brackets `[]`.

In a list we can assign multiple values with comma seperared.

```python
my_list = [90, 25, 36]
print(my_list)
print(type(my_list))
```

```python
my_list2 = ['Ram', 'Shyam']
print(my_list2)
print(type(my_list2))
```

```python
my_list3 = [2.3, 56.2, 589.23]
print(my_list3)
print(type(my_list3))
```

```python
my_list2 = ['Ram', 'Shyam', 2.3, 56.2, 589.23, 59, 23, 3+4j, True, 3==4]
print(my_list2)
print(type(my_list2))
```

```python
my_list = [1, 2, 3, 1, 2, 3, 4, 8, 9, 1, 1, 1, 2, 3]
print(my_list)
print(type(my_list))
```

### 7. Tuples

When we want to assing multiple values to a single variable.

Tuple is defined with the round brackets or paranthesis `()`.

In a tuple we can assign multiple values with comma seperared.

```python
my_tuple = (12, 36, 56)
print(my_tuple)
print(type(my_tuple))
```

```python
my_tuple2 = ('Ram', 'Shyam', 2.3, 56.2, 589.23, 59, 23, 3+4j, True, 3==4)
print(my_tuple2)
print(type(my_tuple2))
```

```python
my_tuple = (1, 2, 3, 1, 2, 3, 4, 8, 9, 1, 1, 1, 2, 3)
print(my_tuple)
print(type(my_tuple))
```

### 8. Set

Store unique objects only (No repetation allowed).

Set is defined with the curly brackets `{}`.

```python
my_set = {1, 2, 3, 1, 2, 3, 4, 8, 9, 1, 1, 1, 2, 3}
print(type(my_set))
print(my_set)
```

### 9. Dictonary

Dictonary is defined with the curly brackets `{}`.

```python
my_dict = {}
print(type(my_dict))
```
