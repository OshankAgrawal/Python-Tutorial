# Type Casting

Type casting/type conversion is the process of converting a value from one data type to another.

There are two types of type conversion in Python.

1. **Implicit Conversion** - automatic type conversion
2. **Explicit Conversion** - manual type conversion

Implicit Conversion:-Python promotes the conversion of the lower data type (integer) to the higher data type (float) to avoid data loss.


In Python, explicit conversion is performed using functions such as int(), float(), str(), and bool(), while implicit conversion can occur automatically when Python safely promotes one compatible type to another.

### Important Type Casting Functions:
    - int()
    - float()
    - str()
    - bool()
    - list()
    - tuple()
    - set()
    - dict()

## 1. int() — Convert to Integer

`int()` converts a value into an integer when possible.

String → Integer

```python
age = "23" 
print(age)
print(type(age))

age = int(age)
print(age)
print(type(age))
```

```python
age = int(input("Enter your age: "))

print(type(age))

print(age)

age += 5

print(age)
```

```python
a = 'hello'

a = int(a)
print(a)
```

```python
a = 'h'
a = int(a)
print(a)
```

Float → Integer

```python
a = 10.8

a = int(a)
print(a)
```

```python
int(-10.8)
```
