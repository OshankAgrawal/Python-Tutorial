# Transfer Statements

## 1. `break` statement

Terminate the current loop. Use the `break` statement to come out of the loop instantly.

Write a python program which print all the natural number until (user enter the number) but it break the loop when the number is 21.

```python
num = int(input("Enter a number"))
print("Entered number is", num)

for i in range(1, num+1):
    if i == 21:
        print("This is break statement.\nWhich terminate the loop.")
        break
    print(i)

print("\nProgram executed successfully.")
```

## 2. `continue` statement

Skip the current iteration of a loop and move to the next iteration

Write a python program which print all the natural number until (user enter the number) but this program does not print a number i.e. 13

```python
num = int(input("Enter a number"))
print("Entered number is:", num)

for i in range(1, num+1):
    
    if i == 13:
        continue

    print(i, end=" ")
```

## 3. `pass` statement

Do nothing. Ignore the condition in which it occurred and proceed to run the program as usual

```python
## pass
## The pass statement is a null operation; it does nothing.

for i in range(5):
    if i==3:
        pass
    print(i)
```

```python
num = 3

match num:
    case 1:
        print("Case 1")
    case 2:
        print("case 2")
    case 3:
        pass ## Pass statement create a empty block.
    case 4:
        print("case 4")
    case _:
        print("Default case")
```
