# List

* The Python lists are containers that are used to store a list of values of any type.
* Unlike other variables Python `lists are mutable` i.e., you can change the elements of a list in place.
* Python will not create a fresh list when you make changes to an element of a list.
* List is a type of sequence like strings and tuples but it differs from them in the way that lists are mutable but strings and tuples are immutable.

## 2. The `append()` method

append( ) method `adds an item to the end of the list`.

Syntax : List.append(<item>)

Note: Takes exactly one element and returns no value

```python
lst = [1, 2, 3, 7, 8, 9, "hello", True]
print(lst)

lst.append('world')

print(lst)
```

```python
lst = [1, 2, 3, 7, 8, 9, "hello", True]

print(id(lst))
print(len(lst))  # 8
print(lst)

lst.append('world')

print(id(lst))
print(len(lst)) # 9
print(lst)
```

## 3. The `insert()` method

The insert( ) method is also an insertion method for lists, like append method.

However, append( ) method add the element at the end of the list.

If you want to `insert an element somewhere in between or any position of your choice`, both append( ) and extend( ) are of no use.

For such a requirement `insert( )` is used.

The `insert( )` function `inserts an item at a given index`.

Syntax: List.insert( <index>, <item>)

Note: Takes two arguments and returns no value.

The first argument <index> is the `index of the element` before which the second argument <item> is to be added.

```python
lst = [1, 2, 3, 7, 8, 9, "hello", True]
print(lst)

lst.insert(3, 'world')

print(lst)
```

```python
lst = [1, 2, 3, 7, 8, 9, "hello", True]

print(id(lst))
print(len(lst))  # 8
print(lst)

lst.insert(3, 'world')

print(id(lst))
print(len(lst)) # 9
print(lst)
```

```python
lst = [1, 2, 3, 7, 8, 9, "hello", True]

lst.insert(len(lst), 'world')

print(lst)
```

```python
lst = [1, 2, 3, 4, 5]

lst.insert(0, 10)

print(lst)
print(len(lst))
```

```python
lst = [1, 2, 3, 4, 5]

lst.insert(-2, 10)

print(lst)
print(len(lst))
```


How it Works

1. Initial List: lst = [1, 2, 3, 4, 5] (Length is 5).
2. Negative Index Resolution: Python converts a negative index by adding it to the length of the list (length + index).
	* 5 + (-2) = 3
3. Insertion: lst.insert(-2, 10) becomes equivalent to lst.insert(3, 10).
4. Shift: The element at index 3 (which is 4) and everything after it shifts to the right, placing 10 right before 4.

Visual Index Breakdown

* lst[0] = 1
* lst[1] = 2
* lst[2] = 3
* lst[3] = 10 (Newly inserted)
* lst[4] = 4
* lst[5] = 5


## 4. The `extend()` method

The `extend( )` method is also used `for adding multiple elements` (given in the form of a list) to a list.

But it is different from append( ). First, let us understand the working of extend( ) function then we'll talk about the difference between these two functions.

The `extend( )` function works as per following format :

Syntax: List.extend(<list>)

Note: Takes exactly one element (`a list type`) and returns no value

That is `extend( ) takes a list as an argument` and `appends all of the elements of the argument list` to the list object on which extend( ) is applied.

```python
lst = [1, 2, 3, 4, 5]

lst.extend([10, 20, 30])

print(lst)
```

### Difference between `append( )` and `extend( )`

5. The `pop()` method

The `pop( )` is used to `remove the item from the list`.

Syntax : List.pop(<index>) 

* <index> is optional argument

Note:  Takes one optional argument and `returns a value - the item being deleted`.

Thus, `pop( )` removes an element from the given position in the list, and return it. 

If no index is specified, pop( ) removes and returns the `last item` in the list.

```python
lst = [1, 2, 3, 4, 5]

ele = lst.pop(2)

print(f"Element {ele}")
print(f"List {lst}")
```

```python
lst = [1, 2, 3, 4, 5]

ele = lst.pop()

print(f"Element {ele}")
print(f"List {lst}")
```

```python
lst = [1, 2, 3, 4, 5]

ele = lst.pop(-2)

print(f"Element {ele}")
print(f"List {lst}")
```

## 6. The `remove()` method

The `remove( )` method `removes` the `first occurrence of given item` from the list.

Syntax: List.remove ( <value> )

Note: Takes one essential argument and does not return anything.

The `remove()` will report `an error if there is no such item in the list`.

```python
lst = [1, 2, 3, 4, 5]

lst.remove(2)

print(f"List {lst}")
```

```python
lst = [1, 2, 3, 4, 5]

lst.remove(6)

print(f"List {lst}")
```

### Difference between `pop()` and `remove()`

* `pop( )` removes an element `whose position is given`, while `remove()` removes an element `whose value is given`.

* `pop()` return the value of element after deleting or removing while `remove()` doesn't return any value.
