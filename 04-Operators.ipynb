# Operators & Their Precedence

Operators are special symbols that perform specific operations on one or more operands (values) and then return a result. 

Types of operator:

1. Arithmatic Operator ---> Numerical Calculations.
    - Addition
    - Substraction
    - Multiplication
    - Division
    - Floor Division
    - Modulus
    - Exponent

2. Assignment Operator ---> Calulate as well as assign the value to the variable.
    - Equal
    - Addition & assign
    - Substraction & assign
    - Multiplication & assign
    - Division & assign
    - Floor Division & assign
    - Modulus & assign
    - Exponent & assign

3. Comparision Operator ---> Compare two or more variables or values and output of these operators are `True` or `False`.
    - greaterthan
    - Lessthan
    - greaterthan equal
    - lessthan equal
    - equal
    - Not equal

4. Logical Operator ---> Logics fininding generally outpus of these operators are `True` or `False`.
    - and
    - or

5. Identity Operator
    - is
    - is not

6. Membership Operator
    - in
    - not in

### 5. Identity Operator

Identity operators are used to compare the objects if both the objects are actually of the same data type and share the same memory location.

Identity operators in Python are `is` and `is not`.

`is` returns `True` when two variables refer to the same object, while `is not` returns `True` when they refer to different objects.

Unlike `==`, which compares values, `is` compares object identity. `is/is not` are commonly used with singleton values such as None.

### 5.1 `is` operator:

It returns `True` when `a` and `b` refer to the **same object**.

The `is` operator returns Boolean either `True` or `False`.

It return `True` if the memory address of the first value is `equal` to the second value ,otherwise it returns `False`.

### 5.2 Is not operator:

`is not` is the opposite of `is`.

returns `True` when `a` and `b` refer to **different objects**.

It returns Boolean values either `True` or `False`.

```python
a = 10
b = 20
print(a is not b)
```

```python
a = 258
b = 258

print(a is not b)
print(id(a))
print(id(b))
```

```python
a = [1, 2, 3, 4]
b = [1, 2, 3, 4]

print(a is b)
```

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a is b)       # False → different objects
print(a is not b)   # True  → different objects

print(a is c)       # True  → same object
print(a is not c)   # False → same object
```

## One Important Comparision

### comparison between `==` ans `is`.

`==` asks: "Are you carrying the same thing?"

`is` asks: "Are you literally the same object?"

```python
a = [1, 2, 3]
b = [1, 2, 3]
c = a

print(a == b)    # True --> Same Value

print(a is b)    # False --> Different Object

print(a is c)    # True ---> Same Object

print(a == c)    # True ---> Same Value 
```

### 6. Membership Operator

Membership operators in Python are `in` and `not in`.

They are used to test whether a **value or element** is **present** in a **sequence or collection such as a string, list, tuple, set, or dictionary**.

`in` returns `True` if the **element exists**, while `not in` returns `True` if it **doesn't exist**.

### 6.1 In Operator:

It returns a result as `True`, if it **finds a given object in the sequence**. Otherwise it returns a `False`.

```python
a = [1, 2, 3, 4, 5]

print(5 in a)
```

```python
a = ()
print(a)
print(id(a))
```

### 6.2 Not in operator:

It returns `True` if the **object is not present in a given sequence** otherwise it returns a `False`.

```python
numbers = [10, 20, 30, 40]

print(50 not in numbers)
print(20 not in numbers)
```

```python
name = "Oshank"

print("o" in name)
print("O" in name)
print("z" in name)
```

```python
text = "I am learning Python."

print("Python" in text)
print("Java" in text)
```

```python
text = "I am learning Python"

print("Java" not in text)
print("Python" not in text)
```

```python
text = "Hello World! this is a new line.Append operation taking place! Append operation taking place! Append operation taking place! First line Seocnd line Third line"

print("this is a newline" in text)
```

```python
fruits = ("apple", "banana", "mango")

print("apple" in fruits)
print("orange" in fruits)
```

```python
numbers = {10, 20, 30, 40}

print(20 in numbers)
print(100 in numbers)
```

## In Case of Dictonary:

When you use `in` with a **dictionary**, Python checks the `keys`, not the `values`.

```python
student = {
    "name": "Oshank",
    "age": 23,
    "city": "Indore"
}

print("name" in student)
print("Oshank" in student)
```

```python
a = student.values()
print(a)
print(type(a))
```

```python
print("Oshank" in student.values())
```

```python
print("name" in student.keys())
```
