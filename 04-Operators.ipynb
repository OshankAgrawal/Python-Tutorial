### 4. Logical Operators

### 4.1 `and` Operator

True only if both the operands are True

```python
print(True and True)
print(True and False)
print(False and True)
```

```python
x = True
y = True

print(x and y)
```

```python
x = True
y = False

print(x and y)
```

```python
x = False
y = True

print(x and y)
```

```python
x = False
y = False

print(x and y)
```

### 4.2 `or` Operator

True if at least one of the operands is True

```python
x = True
y = True

print(x or y)
```

```python
x = True
y = False

print(x or y)
```

```python
x = False
y = True

print(x or y)
```

```python
x = False
y = False

print(x or y)
```

### 4.3 `not` Operator

True if the operand is False and vice-versa.

```python
x = True
print(not x)
```

```python
x = False
print(not x)
```

### Note: In the case of arithmetic values, Logical `and` always return the second value.


```python
35.5 and 4
```

```python
3 and 0
```

```python
0 and 3
```

```python
3 and -4
```

```python
-3 and 0
```

```python
0 and -3
```

### Note: In the case of arithmetic values, Logical `or` it always returns the first value

```python
3 or 4
```

```python
-3 or -4
```

```python
-3 or 0
```

```python
0 or 3
```

### 5. Identity Operator

Identity operators are used to compare the objects if both the objects are actually of the same data type and share the same memory location.

### 5.1 `is` operator:

The `is` operator returns Boolean `True` and `False`.

It return `True` if the memory address of the first value is `equal` to the second value ,otherwise it returns `False`.


```python
a = 10
b = 10

print(a is b)
```

```python
print(id(a))
print(id(b))
```

```python
del (a)
```

```python
print(id(b))
```

```python
print(id(a))
```

```python
a = 10
```

```python
print(id(a))
```

### 5.2 Is not operator:

It returns Boolean values either `True` or `False`.

It return `True` if the first value is `not equal` to the second value otherwise it returns `false`.

```python
a = 10
b = 20
print(a is not b)
```
