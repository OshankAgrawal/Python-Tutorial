# Tuples

The Python tuples are sequences that are used to store a tuple of values of any type.\
Python tuples are immutable i.e., you cannot change the elements of a tuple in place.\
Python will create a fresh tuple when you make changes to an element of a tuple.\
Tuple is a type of sequence like strings and lists but it differs from them in the way that lists are mutable but strings and tuples are immutable.

## Creating tuple

A tuple is a standard data type of Python that can store a sequence of values belonging to an type. The Tuples are depicted through parentheses i.e., round brackets.

Creating a tuple is similar to list creation, but here you need to put a number of expressions in parentheses. That is, use round brackets to indicate the start and end of the tuple, and separate the items by commas.

Thus to create a tuple you can write in the form given below :
T=()
T= (value1, value2, ... )

```python
tup = (1, 2, 3, 4, 5)

print(tup)
print(type(tup))
```

```python
tup1 = (1, 2, 3.0, 4.56, 'hello', 3+4j, True, [1, 2, 3], (10, 20, 30), {89, 66, 36, 89})

print(tup1)

for item in tup1:
    print(f"Item is {item} with data type: {type(item)}")
```

```python
tup = list((1, 2, 3, 4, 5))

print(tup)
print(type(tup))
```

```python
tup = tuple((1, 2, 3, 4, 5))

print(tup)
print(type(tup))
```

## 5. Creating Tuples from Existing Sequences

You can also use the built-in tuple type object (tuple( ) ) to create tuples from sequences as per the syntax given below :

## Operations on Tuple

## 1. Indexing

you can access the tuple elements just like you access a list's or a string's elements e.g. `Tuple[i]` will give you element at i" index.

T[i] returns the item at index i (the first item has index 0),

```python
tup = (0, 1, 4, 9, 16, 25, 36, 49, 64, 81, 100)

print(tup[2])
print(tup[-2])
```

## 2. Slicing

Tuple[a:b] will give you elements between indexes from a to b-1 and so on.

T[i: j] returns a new tuple, containing the objects between i and j excluding index j.

T[i : j : n] returns a new tuple containing every nth item from index i to j, excluding index j. Just like lists.

```python
tup = (0, 1, 4, 9, 16, 25, 36, 49, 64, 81, 100)

print(tup[-2:-7:-1])
```

## 3. `Membership operators`

Both `in` and `not in` operators work on Tuples just like they work for other sequences. That is, it tells if an element is present `in` the tuple or not and `not in` does the opposite.

```python
tup = (0, 1, 4, 9, 16, 25, 36, 49, 64, 81, 100)

print(4 in tup)
```

## 4. Concatination

The + operator, the concatenation operator, when used with two tuples, joins two tuples.

The `+` operator adds one tuple to the end of another.

```python
tup1 = (1, 2, 3, 4, 5)
tup2 = (10, 20, 30, 40, 50)

tup3 = tup1 + tup2

print(tup3)
```

```python
tup1 = (1, 2, 3, 4, 5)
tup2 = (10, 20, 30, 40, 50)
print(id(tup1))
tup1 = tup1 + tup2
print(id(tup1))
print(tup1)
```

## 5. Repetation

The * operator repeats a tuple.

```python
tup1 = (1, 2, 3, 4, 5)

tup3 = tup1*2

print(tup3)
```

```python
tup1 = (1, 2, 3, 4, 5)

tup1 = tup1*2

print(tup1)
```

## 7. `Comparing Tuples`

You can compare two tuples without having to write code with loops for it. For comparing two tuples, you can using comparison operators, i.e., <, >, =. != etc.

For comparison purposes, Python internally compares individual elements of two tuples, applying all the comparison rules that you have read earlier.

```python
tup1 = (1, 2, 3, 4, 5)
tup2 = (10, 20, 30, 40, 50)

for a in range(len(tup1)):
    if tup1[a] != tup2[a]:
        print("Both the tuples are not equal")
        break
else:
    print("Both the tuples are same")
```

```python
tup1 = (1, 2, 3, 4, 5)
tup2 = (10, 20, 30, 40, 50)

print(tup1 == tup2)
```

```python
tup1 = (1, 2, 3, 4, 5)
tup2 = (1, 2, 3, 4, 5, 6)

print(tup1 == tup2)
```

```python
tup1 = (1, 2, 3, 4, 5)
tup2 = (1, 2, 3, 4, 5, 6)

print(tup1 > tup2)
```

## Modification in tuple

Tuples are immutable counterparts of lists. If you need to modify the contents often of a sequence of mixed types, then you should go for lists only.\
However, if you want to ensure that a sequence of mixed types is not accidently changed or modified, you go for tuples.\
But sometimes you need to retain the sequence as tuple and still need to modify the contents at one point of time, then you can use one of the two methods given here.

1. Using tuple unpacking
2. Using the constructor function of list and tuple i.e. list() and tuple()

### 1. Using tuple unpacking

Tuples are immutable. To change a tuple, we would need to first unpack it, change the values, and then again repack it :

```python
# Tuple which you want to modify
tpl = (11, 22, 33, 44, 55)
print(id(tpl))
print(tpl)

# Step 1: First unpack the tuple
a, b, c, d, e = tpl

# Step 2: Modifying or redefine or change the desired variable say, c
c = 77
print(f"Updated value of the changed variable: {c}")

# Step 3: Now again repack the tuple with changed values
tpl = a, b, c, d, e

print(id(tpl))
print(tpl)
```

### 2. Using the constructor of list and tuple i.e. list() and tuple()

```python
# Tuple which you want to modify
tpl = (11, 22, 33, 44, 55)
print(tpl, type(tpl))

# Step 1: Convert the tuple to list using list()
lst = list(tpl)
print(lst, type(lst))

# Step 2: Make change in th desired element in the list
lst[2] = 77

# Step 3: Create  a tuple from the modified list with tuple()
tpl = tuple(lst)
print(tpl, type(tpl))
```
