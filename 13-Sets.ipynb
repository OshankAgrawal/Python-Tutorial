# Sets

Sets are a built-in data type in Python used to store collections of unique items.\
They are unordered, meaning that the elements do not follow a specific order, and they do not allow duplicate elements.\
Sets are useful for membership tests, eliminating duplicate entries, and performing mathematical set operations like union, intersection, difference, and symmetric difference.

## 1. Creating a set

```python
st = {1, 2, 3, 4, 5}

print(st)
print(type(st))
```

## 1.1 Creating empty set

```python
st = set()
print(type(st))
```

```python
st = set([])
print(type(st))
print(st)
```

## 1.2 Creating set from existing sequences

You can also use the built-in set type object (set( ) ) to create set from sequences as per the syntax given below :

`T= set(<sequence>)`

where `<sequence>` can be any kind of sequence object including strings, lists and tuples.

Python creates the individual elements of the set from the individual elements of passed sequence. If you pass in another set, the set function makes a copy.


### 1.2.1 Creating set from a list

```python
lst = [1, 2, 3, 4, 5, 2, 3, 2, 1, 1]

print(lst)
print(type(lst))

st = set(lst)

print(st)
print(type(st))
```

### 1.2.2 Creating set from a tuple

```python
tp = (1, 2, 3, 4, 5, 2, 3, 2, 1, 1)

print(tp)
print(type(tp))

st = set(tp)

print(st)
print(type(st))
```

### 1.2.3 Creating set from a string

```python
str = "Hey! What's happen?"

print(str)
print(type(str))

st = set(str)

print(st)
print(type(st))
```

### 1.2.4 Creating set from key of a dictonary

```python
dis = {1: "2", 2: "3", 3: "4"}

print(type(dis))
print(dis)

st = set(dis)

print(type(st))
print(st)
```

## 2. Set properties

1. Sets are unoredered.
2. Sets are mutuable but frozenset which are immutable
3. Sets are heterogenous.
4. Sets do not allow duplicate values.

### 2.1 Sets are unordered

Unordered means that the items in a set do not have a defined order.

Set items can appear in a different order every time you use them, and cannot be referred to by index or key.

```python
st = {1, 2, 3, 4, 5}

print(st[2])
```

### 2.2 Sets are mutuable but frozenset which are immutable

```python
st = {1, 2, 3, 4, 5}
print(id(st))
print(st)

st.add(7)

print(id(st))
print(st)
```

```python
lst = [1, 2, 3, 4, 5]

st = frozenset(lst)

print(id(st))
print(st)

st.add(7)

print(id(st))
print(st)
```

### 2.3 Sets are heterogenous.

```python
st = {1, 2, 3.5, "Hello", True, 4+3j}
print(st)
```

```python
st = {1, 2, 3.5, "Hello", True, 4+3j, [10, 20, 30], ('first', 3)}
print(st)
```

### 2.4 Sets do not allow duplicate values.

Sets cannot have two items with the same value.

```python
st = {1, 2, 3, 4, 5, 2, 3, 2, 1, 1}

print(st)
print(type(st))
```

## 3. Set Methods

### 3.1 Access Items of set

You cannot access items in a set by referring to an index or a key.\
But you can loop through the set items using a `for loop`, or ask if a specified value is present in a set, by using the `in` keyword.

```python
thisset = {"apple", "banana", "cherry"}

for x in thisset:
  print(x)
```

### 3.1 add element in set

To add one item to a set use the `add()` method.

```python
st = {1, 2, 3, 4, 5}
print(st)
st.add(7)
print(st)
```

### 3.2 Add Sets

To add items from another set into the current set, use the `update()` method.

```python
st = {1, 2, 3, 4, 5}
print(st)
st.update({7, 8, 9})
print(st)
```

### 3.3 Add Any Iterable

The object in the `update()` method does not have to be a set, it can be any iterable object (tuples, lists, dictionaries etc.).

```python
st = {1, 2, 3, 4, 5}
print(st)
st.update([7, 8, 9])
print(st)
```

### 3.4 remove element from set

To remove an element from set, set provides us three different methods:
1. remove() method
2. discard() method
3. pop() method
4. clear() method
5. del keyword

### 3.4.1 `remove()`

To remove an item in a set, use the `remove()` method.

Note: If the item to remove does not exist, `remove()` will raise an error.

```python
st = {1, 2, 3, 4, 5}
print(st)
st.remove(5)
print(st)
```

```python
st = {1, 2, 3, 4}
print(st)
st.remove(5)
print(st)
```

### 3.4.2 `discard()`

To remove an item in a set, use the `discard()` method.

Note: If the item to remove does not exist, `discard()` will NOT raise an error.

```python
st = {1, 2, 3, 4, 5}
print(st)
st.discard(5)
print(st)
```

```python
st = {1, 2, 3, 4}
print(st)
st.discard(5)
print(st)
```

### 3.4.3 `pop()`

You can also use the `pop()` method to remove an item, but `pop()` method will **remove a random item**, so you cannot be sure what item that gets removed.

The return value of the `pop()` method is the removed item.

Note: Sets are unordered, so when using the pop() method, you do not know which item that gets removed.

```python
st = {1, 2, 3, 4, 5}
print(st)
x = st.pop()
print(x)
```

### 3.4.4 `clear()`

The `clear()` method empties the set:

```python
st = {1, 2, 3, 4, 5}
print(st)

st.clear()

print(st)
```

### 3.4.5 `del` keyword

The del keyword will delete the set completely:

```python
st = {1, 2, 3, 4, 5}
print(st)

del st

print(st)
```

## 4. Set Operations

### 4.1 `Union()`

The `union()` method returns a new set with all **items from both sets**.

Note: The union() method allows you to join a set with other data types, like lists or tuples.

```python
# Join set1 and set2 into a new set:

set1 = {"a", "b", "c"}
set2 = {1, 2, 3}

set3 = set1.union(set2)
print(set3)
```

```python
x = {"a", "b", "c"}
y = (1, 2, 3)

z = x.union(y)
print(z)
```

```python
set1 = {"a", "b", "c"}
set2 = {1, 2, 3}
set3 = {"John", "Elena"}
set4 = {"apple", "bananas", "cherry"}

myset = set1.union(set2, set3, set4)
print(myset)
```

### 4.2 `Update()`

The `update()` method inserts all items from one set into another.

The `update()` changes the original set, and does not return a new set.

```python
# The update() method inserts the items in set2 into set1:

set1 = {"a", "b" , "c"}
set2 = {1, 2, 3}

set1.update(set2)
print(set1)
```

### 4.3 Intersection

Keep ONLY the duplicates

The `intersection()` method will return a new set, that only contains the **items that are present in both sets**.

```python
# Join set1 and set2, but keep only the duplicates:

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.intersection(set2)
print(set3)
```

### 4.4 `intersection_update()`

The `intersection_update()` method will also keep ONLY the duplicates, but it will change the original set instead of returning a new set.

```python
# Keep the items that exist in both set1, and set2:

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.intersection_update(set2)

print(set1)
```

The values True and 1 are considered the same value. The same goes for False and 0.\
Join sets that contains the values True, False, 1, and 0, and see what is considered as duplicates:

```python
set1 = {"apple", 1,  "banana", 0, "cherry"}
set2 = {False, "google", 1, "apple", 2, True}

set3 = set1.intersection(set2)

print(set3)
```

### 4.5 `Difference()`

The `difference()` method will return a new set that will contain only the items from the first set that are not present in the other set.

```python
# Keep all items from set1 that are not in set2:

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.difference(set2)

print(set3)
```

```python
set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set2.difference(set1)

print(set3)
```

### 4.6 `difference_update()`

The `difference_update()` method will keep the items from the first set that are not in the other set, but it will change the original set instead of returning a new set.

```python
# Use the difference_update() method to keep only the items from the first set that are not present in the other set:

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.difference_update(set2)

print(set1)
```

### 4.7 `Symmetric Differences`

The `symmetric_difference()` method will keep only the elements that are NOT present in both sets.

```python
# Keep the items that are not present in both sets:

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set3 = set1.symmetric_difference(set2)

print(set3)
```

### 4.8 `symmetric_difference_update()`

The `symmetric_difference_update()` method will also keep all but the duplicates, but it will change the original set instead of returning a new set.

```python
# Use the symmetric_difference_update() method to keep the items that are not present in both sets:

set1 = {"apple", "banana", "cherry"}
set2 = {"google", "microsoft", "apple"}

set1.symmetric_difference_update(set2)

print(set1)
```

### 4.9 `copy()`

The `copy()` method copies the set.

```python
set1 = {"google", "microsoft", "apple"}

set2 = set1.copy()

print(id(set1))
print(id(set2))
print(set2)
```

### 4.10 `isdisjoint()`

The isdisjoint() method returns True if none of the items are present in both sets, otherwise it returns False.\
Return True if no items in set x is present in set y:

```python
x = {"apple", "banana", "cherry"}
y = {"google", "microsoft", "facebook"}

z = x.isdisjoint(y)

print(z)
```

```python
# no common element will give you true

sets_1={1,2,3,4}
sets_2={9,6,7,8}
sets_1.isdisjoint(sets_2)
```

### 4.11 `issubset()`

The issubset() method returns True if all items in the set exists in the specified set, otherwise it returns False.\
Return True if all items in set x are present in set y:

```python
x = {"a", "b", "c"}
y = {"f", "e", "d", "c", "b", "a"}

z = x.issubset(y)

print(z)
```

```python
a={1,3}
b={1,2,3,4}
a.issubset(b)# a ke sare elements b me hona chaiye
#a should be in b
```

### 4.12 `issuperset()`

The issuperset() method returns True if all items in the specified set exists in the original set, otherwise it returns False.\
Return `True` if all items set y are present in set x:

```python
x = {"f", "e", "d", "c", "b", "a"}
y = {"a", "b", "c"}

z = x.issuperset(y)

print(z)
```

```python
a={1,3}
b={1,2,3,4}
a.issuperset(b) #b me a ke sare elements hona chaiye
#b should contain a
```
