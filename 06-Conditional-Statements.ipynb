# Conditional Statements

## 1. if statements

The if decision making statement in Python executes the code, if the condition is true

```python
## if statement
age=20

print("This statement is executed before the if block.")

if age >= 18:
    print("This statement is executed with the if block.")
    print("You are allowed to vote in the elections.")

print("This statement is executed after the if block.")
```

```python
## if statement
age=15

print("This statement is executed before the if block.")

if age >= 18:
    print("This statement is executed with the if block.")
    print("You are allowed to vote in the elections.")

print("This statement is executed after the if block.")
```

## 2. if - else statements

The if…else decision making statement in python executes some code if the condition is true and another code if the condition is false.the false condition comes under else

```python
## else
## The else statement executes a block of code if the condition in the if statement is False.

age=26

print("This statement is executed before the if-else block.")

if age >= 18:
    print("This statement is executed with the if block.")
    print("You are eligible for voting")
else:
    print("This statement is executed with the else block.")
    print("You are a minor")

print("This statement is executed after the if-else block.")
```

```python
## else
## The else statement executes a block of code if the condition in the if statement is False.

age=16

print("This statement is executed before the if-else block.")

if age >= 18:
    print("This statement is executed with the if block.")
    print("You are eligible for voting")
else:
    print("This statement is executed with the else block.")
    print("You are a minor")

print("This statement is executed after the if-else block.")
```

## 3. if - elif - else statements

The if…elif…else statement in python execute code for different conditions.If more than two conditions are true,you can use else for the false condition.

```python
## elif
## The elif statement allows you to check multiple conditions. It stands for "else if"

age=12

print("This statement is executed before the if-elif-else block.")

if age < 13:
    print("This statement is executed with the if block.")
    print("You are a child")

elif age < 18:
    print("This statement is executed with the elif block.")
    print("You are a teenager")

else:
    print("This statement is executed with the else block.")
    print("You are an adult")

print("This statement is executed after the if-elif-else block.")
```

```python
## elif
## The elif statement allows you to check multiple conditions. It stands for "else if"

age=17

print("This statement is executed before the if-elif-else block.")

if age < 13:
    print("This statement is executed with the if block.")
    print("You are a child")

elif age < 18:
    print("This statement is executed with the elif block.")
    print("You are a teenager")

else:
    print("This statement is executed with the else block.")
    print("You are an adult")

print("This statement is executed after the if-elif-else block.")
```

```python
## elif
## The elif statement allows you to check multiple conditions. It stands for "else if"

age=21

print("This statement is executed before the if-elif-else block.")

if age < 13:
    print("This statement is executed with the if block.")
    print("You are a child")

elif age < 18:
    print("This statement is executed with the elif block.")
    print("You are a teenager")

else:
    print("This statement is executed with the else block.")
    print("You are an adult")

print("This statement is executed after the if-elif-else block.")
```

## 4. Nested if - else statements

```python
## Nested Conditonal Statements

# You can place one or more if, elif, or else statements inside another if, elif, or else statement to create nested conditional statements.

## number even ,odd,negative

num=int(input("Enter the number"))

if num > 0:
    print("The number is positive")
    if num%2 == 0:  # num = 12
        print("The number is even")
    else:
        print("The number is odd")

else:
    print("The number is zero or negative")
```

## Example:

```python
### Determine the ticket price based on age and whether the person is a student.
# Ticket pricing based on age and student status

# Take user input
age = int(input("Enter your age: "))
is_student = input("Are you a student? (yes/no): ").lower()

print("Age: ", age)
print("Is student: ", is_student)

# Determine ticket price
if age < 5:
    price = "Free"

elif age <= 12:
    price = "$10"

elif age <= 17:
    if is_student == 'yes':
        price = "$12"
    else:
        price = "$15"

elif age <= 64:
    if is_student == 'yes':
        price = "$18"
    else:
        price = "$25"
        
else:
    price = "$20"

print("Ticket Price:", price)

```

```python
## Assignment
## Simple Calculator program
# Take user input
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")

# Perform the requested operation
if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Error! Division by zero."
else:
    result = "Invalid operation."

print("Result:", result)
```

## Example: User Login System
A simple user login system that checks the username and password.

```python
# User login system

# Predefined username and password
stored_username = "admin"
stored_password = "password123"

# Take user input
username = input("Enter username: ")
password = input("Enter password: ")

# Check login credentials
if username == stored_username:
    if password == stored_password:
        print("Login successful!")
    else:
        print("Incorrect password.")
        
else:
    print("Username not found.")
```
