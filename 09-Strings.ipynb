## 2. Membership Operator

### 2.1 `in` operator

Returns `True` if the substring exist in the super string otherwise return `False`.

```python
str = "ab"
sup_str = "abcdefghi"

print(str in sup_str)
```

### 2.2 `not in ` operator

Returns `True` if the substring does not exist in the super string otherwise return `False`.

```python
str = "ab"
sup_str = "abcdefghi"

print(str not in sup_str)
```

## 2. Traversing

```python
text = "ASCII digits are the characters '0' through '9', which are assigned the numbers 48 through 57 in the ASCII character encoding standard. They are used to represent numerical data in computers."
```

```python
for i in text:
    # print(type(i))
    print(i, end=" ")
```

```python
for i in range(10, 30):
    print(type(i))
    print(text[i])
```
